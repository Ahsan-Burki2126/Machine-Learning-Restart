## Logistic Regression 
is used to predict categories/probabilitis of zero and one and true and false only

### Logistic Regression Equation

The linear part:

$$
z = w_1x_1 + w_2x_2 + \dots + w_nx_n + b
$$

The sigmoid function:

$$
\sigma(z) = \frac{1}{1 + e^{-z}}
$$

Final logistic regression model:

$$
\hat{y} = \sigma(z) = \frac{1}{1 + e^{-(w_1x_1 + w_2x_2 + \dots + w_nx_n + b)}}
$$
s

$$
J(w,b) = -\frac{1}{m}\sum (y\log(\hat y) + (1-y)\log(1-\hat y))
$$


```python
import math 
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
```

```python
x_train = np.array([1, 2, 3, 4, 5, 6]) #study hour + sleep
print(x_train.shape)
y_train = np.array([0, 0, 0, 1, 1, 1] ) # pass or not pass 
print(y_train.shape)
```

```python
#initializing the parameters 
w = 0.0
b = 0.0 
alpha = 0.1 
epochs = 1000

def sigmoid(z):
    return  1/(1+math.exp(-z))
#training loop 
m = len(x_train)
for epoch in range(epochs):
    dw = 0 
    db = 0 
    cost = 0 
    for i in range(m):
        z = w*x_train[i] + b
        y_hat = sigmoid(z)

        #compute gradient
        dw = dw + (y_hat - y_train[i])*x_train[i]
        db = db + (y_hat - y_train[i])

        #compute cost of the sigmoid function or logistic regression 
        eps = 1e-9
        cost += -(y_train[i] * math.log(y_hat + eps) + (1 - y_train[i]) * math.log(1 - y_hat + eps))
    dw = dw/m
    db = db/m 

    #update parameters 
    w = w - alpha*dw
    b = b - alpha*db

    if epoch % 10 == 0:
        print(f"Epoch {epoch}, Cost: {cost:.4f}, w: {w:.4f}, b: {b:.4f}")
        
```

```python
y_pred = [sigmoid(w*x + b) for x in x_train]

plt.scatter(x_train, y_train, marker="x", color="red", label="Training Data")
plt.plot(x_train, y_pred, color="blue", label="Logistic Curve")
plt.xlabel("Study Hours")
plt.ylabel("Pass Probability")
plt.legend()
plt.show()
```

```python
def predict(x):
    prob = sigmoid(w*x + b)
    return 1 if prob >= 0.5 else 0
```
