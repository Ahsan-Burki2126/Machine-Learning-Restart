```python
import numpy as np 
import matplotlib.pyplot as plt
```

we will take the same housing data pints from the previous practice of linear regression 


```python
x_train = np.array([1.0,2.0]) #size in 1000 square feet
y_train = np.array([300.0,500.0]) # price in 10,000 dollars
```

```python
m = x_train.shape[0]
print(m)
w = 20
b =11
f_wb = []
#linear regression formual = f_wb(X) = wx[i] + b

for i in range(m):
    f_wbi = w*x_train[i] + b
    f_wb.append(f_wbi)
print(f_wb)
```

```python
"""
computing cost of the linear regression function 
formula = 1/2m sum(f_wbi - y_train[i])**2

"""
cost_sum = 0 

for i in range(m):
    error = f_wb[i] - y_train[i]
    error_squared = (error**2) / (2*m)
    cost_sum = cost_sum + error_squared
print(f"the final error is : {cost_sum}")
```
