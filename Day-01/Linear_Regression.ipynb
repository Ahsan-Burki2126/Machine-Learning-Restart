Linear Regression Simple


```python
#import libraries
import numpy as np
import matplotlib.pyplot as plt
```

```python
# x_train is size of the house in 1000 sqft
#y_train is the price of the house in 10000's dollar
x_train = np.array([1.0,2.0])
y_train = np.array([300.0,500.0])
```

```python
# m is the number of total training examples 

print(f"total number of training examples are: {m}")
```

```python
"""
Normal equation for w,b 

 w = ( Σ(xi - x̄)(yi - ȳ) ) / Σ(xi - x̄)²
 b = ȳ - w * x̄

"""

mean_x = sum(x_train)/len(x_train)
mean_y = sum(y_train)/len(y_train)

numerator = 0
denominator = 0 

for xi,yi in zip(x_train,y_train):
    numerator += (xi - mean_x)*(yi - mean_y)
    denominator += (xi - mean_x)**2
        

w = numerator/denominator
b = mean_y - (w*mean_x)

print(f"the value of w:{w}")
print(f"the value of b:{b}")

def compute_price(w,b,x):
    m = len(x_train)
    f_wb = np.zeros(m)
    for i in range(m):
        f_wb[i] = w*x[i] + b

    return f_wb     
```

```python
prediction = compute_price(w,b,x_train)
print(prediction)
```

```python
plt.scatter(x_train,y_train,color="red",marker="x")
plt.plot(x_train,prediction)
```
